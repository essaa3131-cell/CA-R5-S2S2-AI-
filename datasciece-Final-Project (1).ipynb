## <mark>The above result show some missing values exist in some columns So let's display them </mark>

```python
#yOUR CODES
missing = gbf.isnull().sum()
missing_pct = (missing / gbf.shape[0] * 100).round(2)
pd.DataFrame({'missing_count': missing, 'missing_pct(%)': missing_pct}).query('missing_count > 0')
```



> ## <mark>The result above shows many missing data </mark>

###  To fix this issue You will fill these values with **the median** for numeric (`member_birth_year`, `start_station_id`, `end_station_id`) and with **a proper placeholder category** for categorical (`start_station_name`, `end_station_name`, `member_gender`)


```python
# trying to fill missing values in my datset using for loop 
numeric_missing_cols = ['start_station_id', 'end_station_id', 'member_birth_year']

for col in numeric_missing_cols:
    gbf[col] = gbf[col].fillna(gbf[col].median())

# the 197 trips with a missing station id also miss the station name (likely a valet / unlisted
# pop-up station), so I flag them with a clear placeholder name instead of guessing a real station
gbf['start_station_name'] = gbf['start_station_name'].fillna('Unknown Station')
gbf['end_station_name'] = gbf['end_station_name'].fillna('Unknown Station')
```

```python
#check the Nulls 
gbf.isnull().sum()
```

### **And with a proper category for other (member gender)**

```python
##  still got a missing value in the member_gender, so let's explore this column more 
gbf.member_gender.value_counts(dropna=False)
```

```python
# fill  dataset with missing values.
gbf['member_gender'] = gbf['member_gender'].fillna('Other')
```

```python
# print(gbf.member_gender.head(15))
gbf.member_gender.value_counts(dropna=False)
```

```python
## check my result above 
gbf.isnull().sum().sum()
```

- We got missing data (mainly in `member_birth_year`/`member_gender`, ~4.5% of rows, and station id/name, ~0.1% of rows), which we fixed by median/placeholder imputation; this is a reasonable approach given the small percentage involved, but it can slightly flatten the true variation in age and station usage.

- The `member_birth_year` column contains clearly invalid values (as old as 1878), producing unrealistic ages (up to 141); these outliers were not removed from the dataset, only excluded visually from some plots, so any deeper statistical modelling would need to filter or cap them first.

- The `start_time` / `end_time` columns in this file only contain minutes:seconds (no date or hour), so no analysis of trip timing by hour-of-day, day-of-week, or weekday/weekend could be performed, even though that would normally be a rich source of insight for a bike-share dataset.
